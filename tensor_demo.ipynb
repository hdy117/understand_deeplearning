```python
import torch
import logging

logging.basicConfig(level=logging.INFO)
logging.info(f"PyTorch version: {torch.__version__}")

def print_separator(info_str:str=None):
    print("#" * 50)
    if info_str:
        print(f"{info_str}")
        print("#" * 50)
```

```python
# different kinds of tensor, rank is ndim, matrix is 2D tensor

# tensor example
print_separator("different kinds of tensor, rank is ndim, matrix is 2D tensor")

tensor_scalar=torch.tensor(3)
print(f'tensor_scalar:{tensor_scalar}, shape:{tensor_scalar.shape}, rank:{tensor_scalar.ndim}')

print_separator()

tensor_scalar = torch.tensor([1,2,3,4,5])
print(f'tensor_scalar:{tensor_scalar}, shape:{tensor_scalar.shape}, rank:{tensor_scalar.ndim}')

print_separator()

tensor_matrix = torch.tensor([[1,2,3],[4,5,6]])
print(f'tensor_matrix:\n\t{tensor_matrix}, \nshape:{tensor_matrix.shape}, rank:{tensor_matrix.ndim}')

print_separator()

tensor_tensor = torch.tensor([[[1,2],[3,4]],[[5,6],[7,8]]])
print(f'tensor_tensor:\n\t{tensor_tensor}, \nshape:{tensor_tensor.shape}, rank:{tensor_tensor.ndim}')
```

```python
# tensor and matrix operations
# the last 2 dims are treated as matrix, the other dims are treated as batch [!!!VERY IMPORTANT!!!]

# 2d tensor or matrix multiplication
print_separator(f"the last 2 dims are treated as matrix, the other dims are treated as batch [!!!VERY IMPORTANT!!!]")

tensor_matrix1 = torch.tensor([[1,2,3],[4,5,6]])
tensor_matrix2 = torch.tensor([[7,8],[9,10],[11,12]])
tensor_matrix_product = torch.matmul(tensor_matrix1, tensor_matrix2)
print(f'tensor_matrix1:\n\t{tensor_matrix1}, \nshape:{tensor_matrix1.shape}, rank:{tensor_matrix1.ndim}')
print(f'tensor_matrix2:\n\t{tensor_matrix2}, \nshape:{tensor_matrix2.shape}, rank:{tensor_matrix2.ndim}')
print(f'tensor_matrix_product:\n\t{tensor_matrix_product}, \nshape:{tensor_matrix_product.shape}, rank:{tensor_matrix_product.ndim}')

print_separator()

# 3d tensor or batch of matrices multiplication
tensor_x=torch.tensor([[[1,2,3],[4,5,6]],[[1,2,3],[4,5,6]]]) # batch:2, matrix:2x3
tensor_w=torch.tensor([[7,8],[9,10],[11,12]]) # matrix:3x2
tensor_y=torch.matmul(tensor_x, tensor_w) # batch:2, matrix:2x2
print(f"tensor_x:{tensor_x}, tensor_x.shape:{tensor_x.shape}, tensor_x.ndim:{tensor_x.ndim}")
print(f"tensor_w:{tensor_w}, tensor_w.shape:{tensor_w.shape}, tensor_w.ndim:{tensor_w.ndim}")
print(f"tensor_y:{tensor_y}, tensor_y.shape:{tensor_y.shape}, tensor_y.ndim:{tensor_y.ndim}")
```

```python
# pytorch tensor, pytorch store W tensor in [D_out, D_in] format, but in deep learning, we usually store W tensor in [D_in, D_out] format, so we need to transpose the W tensor before multiplication
print_separator("pytorch tensor, pytorch store W tensor in [D_out, D_in] format, but in deep learning, we usually store W tensor in [D_in, D_out] format, so we need to transpose the W tensor before multiplication")

Batch=2
D_in=10
D_out=5
tensor_x = torch.randint(1,10,(Batch, D_in)).to(torch.float32) # [Batch, D_in]
tensor_w = torch.randn(size=[D_out, D_in]) # [D_out, D_in]
tensor_y=tensor_x@tensor_w.T # [Batch, D_out]
print(f"tensor_x:{tensor_x}, tensor_x.shape:{tensor_x.shape}, tensor_x.ndim:{tensor_x.ndim}")
print(f"tensor_w:{tensor_w}, tensor_w.shape:{tensor_w.shape}, tensor_w.ndim:{tensor_w.ndim}")
print(f"tensor_y:{tensor_y}, tensor_y.shape:{tensor_y.shape}, tensor_y.ndim:{tensor_y.ndim}")
print_separator()
```

```python
# CNN does not do matrix multiplication, It does convolution, but the convolution can also be implemented as dot product when convolution and masked element are flattened, so the convolution can also be implemented as matrix multiplication, but the convolution is more efficient than matrix multiplication, because the convolution can reuse the masked element, but the matrix multiplication cannot reuse the masked element, so the convolution is more efficient than matrix multiplication.
print_separator("# CNN does not do matrix multiplication, It does convolution, but the convolution can also be implemented as dot product when convolution and masked element are flattened")

# CNN
B, C_in, D_in, C_out = 2, 3, 10, 6
tensor_x = torch.randn(size=[B, C_in, D_in]) # [Batch, C_in, D_in]
conv_layer = torch.nn.Conv1d(in_channels=C_in, out_channels=C_out, kernel_size=3, padding=1, stride=1) # [Batch, C_out, D_out]
tensor_y = conv_layer(tensor_x)
print(f"tensor_x:{tensor_x}, tensor_x.shape:{tensor_x.shape}, tensor_x.ndim:{tensor_x.ndim}")
print(f"tensor_y:{tensor_y}, tensor_y.shape:{tensor_y.shape}, tensor_y.ndim:{tensor_y.ndim}")
print_separator()
```

```python
# transformer example
print_separator("transformer example, softmax(q@k_t/sqrt(d_k))@v")

# batch，sequence length, embedding dimension
B,S,D = 2,3,6
q=torch.randn(size=[B,S,D]) # [Batch, Sequence length, Embedding dimension]
k=torch.randn(size=[B,S,D]) # [Batch, Sequence length, Embedding dimension
v=torch.randn(size=[B,S,D]) # [Batch, Sequence length, Embedding dimension]
print(f"q.shape:{q.shape}, q.ndim:{q.ndim}")
print(f"k.shape:{k.shape}, k.ndim:{k.ndim}")
print(f"v.shape:{v.shape}, v.ndim:{v.ndim}")

# attention score
attention_score = torch.matmul(q,k.transpose(-1,-2))/torch.sqrt(torch.tensor(D)) 
print(f"attention_score.shape:{attention_score.shape}, attention_score.ndim:{attention_score.ndim}")
print(f"attention_score:{attention_score}")

attention_score_softmax = torch.softmax(attention_score, dim=-1)

print(f"attention_score_softmax.shape:{attention_score_softmax.shape}, attention_score_softmax.ndim:{attention_score_softmax.ndim}")
print(f"attention_score_softmax:{attention_score_softmax}")

print_separator()

# routing the attention score to the value
output = torch.matmul(attention_score_softmax, v)
print(f"output.shape:{output.shape}, output.ndim:{output.ndim}")    

```

```python
# self-attention example
print_separator("self-attention example, softmax(q@k_t/sqrt(d_k))@v")

# batch, n_head, n_token, token_len, d_embedding
batch, n_head, n_token, d_embedding = 2, 3, 4, 24
d_head = d_embedding // n_head
sqrt_d_head = torch.sqrt(torch.tensor(d_head, dtype=torch.float32))

x=torch.randn(size=[batch, n_token, d_embedding]) # [Batch, n_token, d_embedding]

# weight matrix for q, k, v, from d_embedding to d_embedding
w_q = torch.randn(size=[d_embedding, d_embedding]) # [d_embedding, D_embedding]
w_k = torch.randn(size=[d_embedding, d_embedding]) # [d_embedding, D_embedding]
w_v = torch.randn(size=[d_embedding, d_embedding]) # [d_embedding, D_embedding]

# compute q, k, v
q = torch.matmul(x, w_q).view(batch, n_token, n_head, d_head).transpose(2, 1) # [Batch,  n_head, n_token, d_head]
k = torch.matmul(x, w_k).view(batch, n_token, n_head, d_head).transpose(2, 1) # [Batch,  n_head, n_token, d_head]
v = torch.matmul(x, w_v).view(batch, n_token, n_head, d_head).transpose(2, 1) # [Batch,  n_head, n_token, d_head]

print(f"q.shape:{q.shape}, q.ndim:{q.ndim}")
print(f"k.shape:{k.shape}, k.ndim:{k.ndim}")
print(f"v.shape:{v.shape}, v.ndim:{v.ndim}")

print_separator()

# attention score before softmax
attention_score = torch.matmul(q,k.transpose(-1,-2))/sqrt_d_head
# attention score after softmax
attention_score = torch.softmax(attention_score, dim=-1)
print(f"attention_score.shape:{attention_score.shape}, attention_score.ndim:{attention_score.ndim}")

# routing the attention score to the value
router_output = torch.matmul(attention_score, v)
print(f"router_output.shape:{router_output.shape}, router_output.ndim:{router_output.ndim}")
print_separator()

```

```python
# cross attention example
print_separator("cross attention example, softmax(q@k_t/sqrt(d_k))@v")

# attention meta size info
batch, n_token, d_embedding = 2,4,24
n_head = 3
head_embedding = d_embedding // n_head
sqrt_head_embedding = torch.sqrt(torch.tensor(head_embedding, dtype=torch.float32))

# input x from query for encoder, x1 from key&value for decoder
x=torch.randn(size=[batch, n_token, d_embedding]) # [Batch, n_token, d_embedding] # 中文输入，encoder部分
x1=torch.randn(size=[batch, n_token, d_embedding]) # [Batch, n_token, d_embedding] # 英文输入，decoder部分

# weight matrix for q, k, v, from d_embedding to d_embedding
w_q = torch.randn(size=[d_embedding, d_embedding])
w_k = torch.randn(size=[d_embedding, d_embedding])
w_v = torch.randn(size=[d_embedding, d_embedding])

# compute q, k, v, and query from x, key and value from x1
q=torch.matmul(x,w_q).view(batch, n_token, n_head, head_embedding).transpose(2,1) # [Batch, n_head, n_token, head_embedding]，encoder部分
k=torch.matmul(x1,w_k).view(batch, n_token, n_head, head_embedding).transpose(2,1) # [Batch, n_head, n_token, head_embedding]，decoder部分
v=torch.matmul(x1,w_v).view(batch, n_token, n_head, head_embedding).transpose(2,1) # [Batch, n_head, n_token, head_embedding]，decoder部分

print(f"q.shape:{q.shape}, q.ndim:{q.ndim}")
print(f"k.shape:{k.shape}, k.ndim:{k.ndim}")
print(f"v.shape:{v.shape}, v.ndim:{v.ndim}")

# attention score
attention_score = torch.matmul(q,k.transpose(-1,-2))/sqrt_head_embedding
attention_score=torch.softmax(attention_score, dim=-1)
print(f"attention_score.shape:{attention_score.shape}, attention_score.ndim:{attention_score.ndim}")

# value
route_output = torch.matmul(attention_score, v)
print(f"route_output.shape:{route_output.shape}, route_output.ndim:{route_output.ndim}")
```

```python
# casual transformer example
# refer to https://waylandz.com/blog/zh/transformer-architecture/

print_separator("casual transformer example, softmax(q@k_t/sqrt(d_k))@v, with mask. refer to https://waylandz.com/blog/zh/transformer-architecture/")

# meta info
batch, n_token, d_embedding = 2,4,24
n_head=3
head_embedding = d_embedding // n_head
sqrt_head_embedding = torch.sqrt(torch.tensor(head_embedding, dtype=torch.float32))

# x, x1, x2 are the input for query, key and value respectively
x=torch.randn(size=[batch, n_token, d_embedding]) # [Batch, n_token, d_embedding]
x1=torch.randn(size=[batch, n_token, d_embedding]) # [Batch, n_token, d_embedding]

# projection
w_q=torch.randn(size=[d_embedding, d_embedding])
w_k=torch.randn(size=[d_embedding, d_embedding])
w_v=torch.randn(size=[d_embedding, d_embedding])

# q, k, v
q=torch.matmul(x,w_q).view(batch, n_token, n_head, head_embedding).transpose(2,1) # [Batch, n_head, n_token, head_embedding]
k=torch.matmul(x,w_k).view(batch, n_token, n_head, head_embedding).transpose(2,1) # [Batch, n_head, n_token, head_embedding]
v=torch.matmul(x,w_v).view(batch, n_token, n_head, head_embedding).transpose(2,1) # [Batch, n_head, n_token, head_embedding]

# attention score
attention_score=torch.matmul(q,k.transpose(-1,-2))/sqrt_head_embedding # [Batch, n_head, n_token, head_embedding]
print(f"attention_score.shape:{attention_score.shape}, attention_score.ndim:{attention_score.ndim}")
# print(f"attention_score:{attention_score}")

# casual mask attention， diagonal=1 means shift the mask to the right by 1, so that the current token can only attend to the previous tokens, and not the future tokens
casual_mask=torch.triu(torch.ones(size=[n_token, n_token]), diagonal=1).bool() # [n_token, n_token]
attention_score = attention_score.masked_fill(casual_mask, float('-inf')) # casual mask attention, set the upper triangle to -inf

# softmax
attention_score = torch.softmax(attention_score, dim=-1)
print(f"attention_score.shape:{attention_score.shape}, attention_score.ndim:{attention_score.ndim}")
print(f"attention_score:{attention_score}")

# routing the attention score to the value
route_output = torch.matmul(attention_score, v)
print(f"route_output.shape:{route_output.shape}, route_output.ndim:{route_output.ndim}")


```

```python
# key value cache example, for decoder model

print_separator("key value cache example, for decoder model")

# meta info
N_batch = 2
d_embedding = 24
n_head = 3
head_embedding = d_embedding // n_head
sqrt_head_embedding = torch.sqrt(torch.tensor(head_embedding, dtype=torch.float32))

# projection weight matrix for q, k, v
w_q=torch.randn(size=[d_embedding, d_embedding])
w_k=torch.randn(size=[d_embedding, d_embedding])
w_v=torch.randn(size=[d_embedding, d_embedding])

# projection, n_token use x.shape[1] to adapt to different length of input, for example, in decoder model, the input is a single token, so n_token=1, but in encoder model, the input is a sequence of tokens, so n_token>1
def projection(x:torch.Tensor)->tuple[torch.Tensor, torch.Tensor, torch.Tensor]:
    n_batch, n_token=x.shape[0], x.shape[1]
    q=torch.matmul(x, w_q).view(n_batch, n_token, n_head, head_embedding).transpose(2,1) # [Batch, n_head, n_token, head_embedding]
    k=torch.matmul(x, w_k).view(n_batch, n_token, n_head, head_embedding).transpose(2,1) # [Batch, n_head, n_token, head_embedding]
    v=torch.matmul(x, w_v).view(n_batch, n_token, n_head, head_embedding).transpose(2,1) # [Batch, n_head, n_token, head_embedding]
    return q,k,v

# kv cache 
N_steps = 10 # suppose decoder model has 10 steps, and we will cache the key and value for each step

all_x =torch.randn(size=[N_batch, N_steps, d_embedding])

# key and value cache
k_cache = None
v_cache = None

# simulate the decoder model, for each step, we will cache the key and value, and use the cached key and value for the next step
for t in range(N_steps):
    x = all_x[:, t, :].unsqueeze(1)  # [Batch, 1, d_embedding]
    q,k,v=projection(x) # [Batch, n_head, 1, head_embedding]
    print(f"t:{t}")
    print(f"q.shape:{q.shape}, k.shape:{k.shape}, v.shape:{v.shape}")
    
    if k_cache is None:
        k_cache=k
        v_cache=v
    else:
        k_cache=torch.cat([k_cache, k], dim=2) # [batch, n_head, t+1, head_embedding]
        v_cache=torch.cat([v_cache, v], dim=2) #[batch, n_head, t+1,head_embedding]
    
    print(f"k_cache.shape:{k_cache.shape}, v_cache.shape:{v_cache.shape}")
    
    # attention score
    attention_score = torch.matmul(q,k_cache.transpose(-1,-2))/sqrt_head_embedding
    attention_score=torch.softmax(attention_score, dim=-1) # [batch, n_head, 1, t+1]
    
    #routing to output
    route_output=torch.matmul(attention_score, v_cache) # [batch, n_head, 1, head_embedding]
```
